```python
import pandas as pd
import requests

# Raw URL for the Excel file on GitHub
url = 'https://raw.githubusercontent.com/ImNotDanish05/2026_Pembelajaran-Mesin_TI-3A/main/Week_11/Time_Series/luas-karhutla-provinsi.xlsx'

# Download the file
response = requests.get(url)
with open('luas-karhutla-provinsi.xlsx', 'wb') as f:
    f.write(response.content)

# Load the Excel file
df = pd.read_excel('luas-karhutla-provinsi.xlsx')

# Print structure information
print("=== DataFrame Info ===")
df.info()

print("\n=== First 5 Rows ===")
display(df.head())
```
